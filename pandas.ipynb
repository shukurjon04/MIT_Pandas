```python
import pandas as p
```

```python
data = {
    "Ism":["Shukurjon","Yusuf","Shomatjon","Nomonjon","Kamronjon"],
    "Familya":["Boqiyev","Zunnunov","Nishonov","Valiyev","Isaqov"],
    "Age":[21,21,22,24,23],
    "Grade":[5,3,4,5,4]
}

baza = p.DataFrame(data)

baza.to_csv("students.csv")
print(baza)
```

```python
baz = p.read_csv("students.csv")

print(baz.head(3))
```

```python
meean = baza['Age'].mean

maxx = baza['Age'].max

minn = baza['Age'].min

print(meean)
print(maxx)
print(minn)
```

```python
gradefiltered = baza[baza['Grade'] > 4]

print(gradefiltered)
```

```python
baza['Condition'] = baza["Grade"].apply(lambda x: "O'tgan" if x >= 4 else "O'tmagan")

print(baza)
print(type(baza))
```

```python
groupedbaza = baza.groupby('Age')

print(groupedbaza['Age'].mean())
```

```python
sorted = baza.sort_values(by='Grade')

print(sorted)
```
